```python
mtxfiles = ! ls ../../../Data/Revision/CRC_Outcome/Idvfiles/*_matrix.mtx.gz
adatas = []
for fn in mtxfiles:
    mat = mmread(fn)
    barcodes = pd.read_csv(fn.replace('matrix.mtx', 'barcodes.tsv'), sep='\t', header=None)
    features = pd.read_csv(fn.replace('matrix.mtx', 'features.tsv'), sep='\t', header=None)
    features.columns=['Ensemble', 'Gene_Symbol', 'feature_types']
    features.index = features['Ensemble'].values
    barcodes['file'] = fn
    barcodes['GSM'] = fn.split('/')[-1].split('_')[0]
    barcodes['MGI'] = fn.split('/')[-1].split('_')[2]=='MGI'
    barcodes['sample'] = fn.split('/')[-1].replace('_MGI', '').split('_')[2]
    barcodes['version'] = fn.split('/')[-1].replace('_MGI', '').split('_')[3]
    barcodes.index = barcodes['sample'] + '_' + barcodes[0]
    adata = sc.AnnData(mat.T, obs=barcodes, var=features)
    adatas.append(adata)
    
```

```python
adata = anndata.concat(adatas)
```

```python
merged = pd.merge(left=adata.obs, right=sample_info.drop('GSM', axis=1), on='sample', how='left')
(adata.obs[0] == merged[0].values).value_counts()
```
